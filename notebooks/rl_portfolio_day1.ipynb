```python
# 测试环境


# 加载少量股票数据测试环境
prices = pd.read_csv('D:/desktop/QuantProject2/data/hs300_prices.csv', index_col=0, parse_dates=True)

# 选择5只股票
test_prices = prices.iloc[:, :5]   # 取前5列

# 创建环境
env = PortfolioTradingEnv(test_prices)

# 重置环境
obs, _ = env.reset()
print("初始观测维度:", obs.shape)
print("初始观测前几个元素:", obs[:6])

# 随机动作测试
for step in range(10):
    action = env.action_space.sample()  # 随机生成一个动作
    obs, reward, done, truncated, info = env.step(action)
    print(f"Step {step}: reward={reward:.5f}, portfolio_value={env.portfolio_value:.4f}, done={done}")
    if done:
        break
```
